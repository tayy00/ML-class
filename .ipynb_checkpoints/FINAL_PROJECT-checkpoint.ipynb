```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, Lasso
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, IsolationForest
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from imblearn.over_sampling import SMOTE
from imblearn.ensemble import BalancedRandomForestClassifier
import shap
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
# Load data (features from secom.data, labels from secom_labels.data)
df = pd.read_csv('secom.data', sep=' ', header=None)
labels = pd.read_csv('secom_labels.data', sep=' ', header=None)[0].map({-1: 0, 1: 1})  # Map -1 to 0 (pass), 1 to 1 (fail)
data = pd.concat([df, labels.rename('label')], axis=1)
```

```python
# Descriptive stats
desc = df.describe().T
desc['variance'] = df.var()
desc['missingness_%'] = df.isna().mean() * 100

# Sample output (full table in appendix)
print(desc.head(10))
```

```python
# ======================
# 1(a) DESCRIPTIVE STATISTICS
# ======================

print("Dataset Shape:", df.shape)

means = df.mean(skipna=True)
variances = df.var(skipna=True)
missing_counts = df.isna().sum()
missing_percent = (missing_counts / len(df)) * 100

# Summary DataFrame
summary_stats = pd.DataFrame({
    "Mean": means,
    "Variance": variances,
    "Missing_Count": missing_counts,
    "Missing_Percent": missing_percent
})


print(summary_stats.head(10))
```

```python
# Distribution of sample feature (Column 0)
plt.figure(figsize=(6,4))
sns.histplot(df[0], kde=True)
plt.title('Distribution of Sensor 0')
plt.show()

# Correlation heatmap (sample 20 features)
corr = df.iloc[:, :20].corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=False, cmap='coolwarm')
plt.title('Correlation Heatmap (First 20 Features)')
plt.show()

# PCA plot
scaler = StandardScaler()
pca = PCA(n_components=2)
pca_components = pca.fit_transform(scaler.fit_transform(df.fillna(df.mean())))
plt.figure(figsize=(6,4))
sns.scatterplot(x=pca_components[:,0], y=pca_components[:,1], hue=labels)
plt.title('PCA Plot of First Two Components')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.show()
```

```python
# 1. Distribution of a few predictors
plt.figure(figsize=(12, 6))
for i, col in enumerate([0, 1, 2, 3]):  # first 4 features
    plt.subplot(2, 2, i+1)
    sns.histplot(df[col].dropna(), bins=30, kde=True)
    plt.title(f"Distribution of Feature {col}")
plt.tight_layout()
plt.show()

# 2. Correlation heatmap (subset of first 20 features)
corr = df.iloc[:, :20].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap (First 20 Features)")
plt.show()

# 3. PCA visualization (first 50 features, scaled, with labels)
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df.iloc[:, :50].fillna(df.mean()))
pca = PCA(n_components=2)
pca_result = pca.fit_transform(df_scaled)

plt.figure(figsize=(8, 6))
sns.scatterplot(x=pca_result[:, 0], y=pca_result[:, 1], hue=labels, palette="Set1", alpha=0.7)
plt.title("PCA Plot (First 50 Features)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="Class (0=Pass, 1=Fail)")
plt.show()

print("\nExplained Variance by PC1 & PC2:", pca.explained_variance_ratio_)
```

# Q2 Predictive Models (Baseline)

```python
#Logistic Regression.
# Preprocess: remove constants, impute mean, scale
constant_cols = df.columns[df.nunique() <= 1]
X = df.drop(constant_cols, axis=1).fillna(df.mean())
y = labels
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_scaled, y_train)
lr_pred = lr.predict(X_test_scaled)


rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

def evaluate(y_true, y_pred):
    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred),
        'Recall': recall_score(y_true, y_pred),
        'ROC AUC': roc_auc_score(y_true, y_pred)
    }

print('LR:', evaluate(y_test, lr_pred))
print('RF:', evaluate(y_test, rf_pred))
```

```python
# ==============================
# QUESTION 2(a): Logistic Regression
# ==============================

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score, classification_report

# --- Preprocessing ---
# Fill missing values with column mean
df_filled = df.fillna(df.mean())

# Standardize features (important for Logistic Regression)
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_filled)

# --- Train-test split ---
X_train, X_test, y_train, y_test = train_test_split(
    df_scaled, y, test_size=0.3, random_state=42, stratify=y
)

# --- Logistic Regression Model ---
log_reg = LogisticRegression(max_iter=500, class_weight="balanced")  
log_reg.fit(X_train, y_train)

# --- Predictions ---
y_pred = log_reg.predict(X_test)
y_prob = log_reg.predict_proba(X_test)[:, 1]

# --- Evaluation Metrics ---
print("Logistic Regression Performance:")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("ROC AUC:", roc_auc_score(y_test, y_prob))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

```

```python
# ==============================
# QUESTION 2(b): Random Forest
# ==============================

from sklearn.ensemble import RandomForestClassifier

# --- Random Forest Model ---
rf_model = RandomForestClassifier(
    n_estimators=200,       # number of trees
    max_depth=None,        # let trees grow deep
    class_weight="balanced",# handle imbalance
    random_state=42,
    n_jobs=-1               # use all CPU cores
)

rf_model.fit(X_train, y_train)

# --- Predictions ---
y_pred_rf = rf_model.predict(X_test)
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

# --- Evaluation Metrics ---
print("Random Forest Performance:")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall:", recall_score(y_test, y_pred_rf))
print("ROC AUC:", roc_auc_score(y_test, y_prob_rf))
print("\nClassification Report:\n", classification_report(y_test, y_pred_rf))
```

```python
# ==============================
# QUESTION 2(c): Model Comparison
# ==============================

# Collect results into a DataFrame for easy comparison
results = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy": [
        accuracy_score(y_test, y_pred),
        accuracy_score(y_test, y_pred_rf)
    ],
    "Precision": [
        precision_score(y_test, y_pred),
        precision_score(y_test, y_pred_rf)
    ],
    "Recall": [
        recall_score(y_test, y_pred),
        recall_score(y_test, y_pred_rf)
    ],
    "ROC_AUC": [
        roc_auc_score(y_test, y_prob),
        roc_auc_score(y_test, y_prob_rf)
    ]
})

print("\n=== Model Comparison ===\n")
print(results)

# Optional: Bar plot for visual comparison
results_melted = results.melt(id_vars="Model", var_name="Metric", value_name="Score")

plt.figure(figsize=(10, 6))
sns.barplot(x="Metric", y="Score", hue="Model", data=results_melted)
plt.title("Model Performance Comparison: Logistic Regression vs Random Forest")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.show()
```
